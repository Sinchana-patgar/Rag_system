EXTRACTING TEXTS FROM PDFS


```python
from langchain_community.document_loaders import PyPDFLoader

loader=PyPDFLoader("./Docs/fabric onelake.pdf")

docs=loader.load()
docs
```

Changing metadata

```python
for i in docs:
    i.metadata={"source":"fabric onelake.pdf","developer":"Microsoft"}
```

```python
docs[0].metadata
```

```python
#step 2: Chunking

from langchain_text_splitters import RecursiveCharacterTextSplitter

splitter= RecursiveCharacterTextSplitter(
chunk_size=1000,
chunk_overlap=100)

chunks=splitter.split_documents(docs)
chunks
```

```python
chunks[0].metadata
```

```python
#embeddings

from langchain_huggingface import HuggingFaceEmbeddings

embeddings=HuggingFaceEmbeddings(model_name="all-MiniLM-L6-v2")
```

```python
import platform

# Force Chroma to use the SegmentAPI (Python) instead of Rust on Windows
if platform.system() == "Windows":
    from chromadb.config import Settings
    # This tells Chroma to ignore the buggy Rust bindings
    settings = Settings(chroma_api_impl="chromadb.api.segment.SegmentAPI")
else:
    settings = None

# Then when you create your vectorstore, pass the settings:
# vectorstore = Chroma.from_documents(documents=splits, embedding=embeddings, client_settings=settings)
```

```python
#vectorstores

from langchain_community.vectorstores import Chroma
persist_directory = "D:/RAG/Vector"
vectorstores = Chroma.from_documents(
    documents=chunks,
    embedding=embeddings,
    persist_directory=persist_directory
)
```

```python
#semantic search
#semantic search
vectorstores.similarity_search("Why do we need one lake",k=3)
```
